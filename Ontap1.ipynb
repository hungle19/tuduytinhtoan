1: Yêu cầu. Cước phí Taxi Thông Minh.
Viết một hàm duy nhất smart_taxi_fare(car_type, start_hour, km, wait_minutes) tính và trả về total (tổng tiền phải trả,
kiểu số nguyên, đơn vị VND).
Các quy tắc cần cài đặt:
1. Giá mở cửa (0–0.8 km):
- Xe 4 chỗ: 11,000 VND
- Xe 7 chỗ: 12,000 VND
2. Giá theo km sau 0.8 km:
- Xe 4 chỗ: từ 0.8→20 km ơnh 13,000 VND/km; trên 20 km tính 10,500 VND/km
- Xe 7 chỗ: từ 0.8→20 km ơnh 14,000 VND/km; trên 20 km tính 12,000 VND/km
3. Phí chờ: 5 phút đầu miễn phí; sau đó, mỗi 5 phút (làm tròn lên theo từng block 5 phút) tính 2,000 VND.
4. Phụ thu ban đêm: nếu start_hour thuộc [22,23,0,1,2,3,4,5], cộng +20% trên tiền tạm tính (mở cửa + quãng đường +
chờ).
5. Điều kiện đặc biệt:
- Nếu km > 30 và không phải ban đêm, áp dụng giảm 10% trên tổng tiền.
- Nếu tổng tiền cuối cùng < 50,000 nhưng km > 5, đặt tổng tiền bằng 50,000 VND (mức tối thiểu). 


```python
import math

def smart_taxi_fare(car_type, start_hour, km, wait_minutes):
    total = 0

    if car_type == 4:
        open_price = 11000
        price1 = 13000
        price2 = 10500
    elif car_type == 7:
        open_price = 12000
        price1 = 14000
        price2 = 12000
    else:
        return 0

    if km <= 0.8:
        total = open_price
    elif km <=20:
        total = open_price + (km - 0.8) * price1
    else:
        total = open_price
        total += (20 - 0.8) * price1
        total += (km - 20) * price2

    if wait_minutes > 5:
        extra_wait = wait_minutes - 5
        blocks = math.ceil(extra_wait / 5)
        total += blocks * 2000

    night = start_hour in [22, 23, 0, 1, 2, 3, 4, 5]
    if night:
        total *= 1.2
    if km > 30 and not night:
        total *= 0.9

    if total < 50000 and km > 5:
        total = 50000

    return int(total)

car_type, start_hour, km, wait_minutes = map(int, input().split())
print(smart_taxi_fare(car_type, start_hour, km, wait_minutes))

```

4: Yêu cầu: Cho các hệ số thực a, b, c, trả về tất cả các nghiệm thực của phương trình bậc hai axx + bx + c = 0 trong một danh
sách được sắp xếp tăng dần. 
Quy tắc:
- Nếu a == 0 và b != 0: trả về nghiệm duy nhất [-c / b].
- Nếu a == 0 và b == 0: trả về [] (không có nghiệm hữu hạn duy nhất nếu c == 0, hoặc vô nghiệm nếu c != 0).
- Với a != 0, dùng biệt thức d = bb - 4ac:
- d < 0 → []
- d == 0 → [ -b / (2a) ]
- d > 0 → hai nghiệm [r1, r2] với r1 <= r2

```python
import math

def solve_quadratic(a, b, c):
    if a == 0:
        if b != 0:
            return [-c / b]
        else:
            return []

    d = b * b - 4 * a * c

    if d < 0:
        return []

    elif d == 0:
        return [-b / (2 * a)]

    else:
        r1 = (-b - math.sqrt(d)) / (2 * a)
        r2 = (-b + math.sqrt(d)) / (2 * a)

        return sorted([r1, r2])

a, b, c = map(float, input().split())
print(solve_quadratic(a, b, c))
```

14: Viết hàm trả về hệ số nhị thức C(n, k). 

```python
def factorial(n):
    result = 1
    for i in range(1, n+1):
        result *= i
    return result
def C(n, k):
    return factorial(n) // (factorial(k) * factorial(n-k))
n, k = map(int, input().split())
print(C(n, k))
```

dictionary) 

```python
def max_min(a, b, c):
    nums = [a, b, c]
    max_val = nums[0]
    min_val = nums[0]
    for x in nums:
        if x > max_val:
            max_val = x
        if x < min_val:
            min_val = x
    return [max_val, min_val]
a, b, c = map(int, input().split())
print(max_min(a, b, c))
```

16: Viết hàm trả về góc ∠ABC theo đơn vị độ khi biết tọa độ các điểm A(x1,y1), B(x2,y2), C(x3,y3). 


```python
import math
def angle_ABC(x1, y1, x2, y2, x3, y3):
    BAx = x1 - x2
    BAy = y1 - y2
    BCx = x3 - x2
    BCy = y3 - y2
    dot = BAx * BCx + BAy * BCy
    len_BA = math.sqrt(BAx**2 + BAy**2)
    len_BC = math.sqrt(BCx**2 + BCy**2)
    cos_angle = dot / (len_BA * len_BC)
    cos_angle = max(-1, min(1, cos_angle))
    angle = math.degrees(math.acos(cos_angle))
    return angle
x1, y1, x2, y2, x3, y3 = map(int, input().split())
print(angle_ABC(x1, y1, x2, y2, x3, y3))
```

trên biên là ở trong. 

```python
def point_in_quadrilateral(quad, p):
    x, y = p
    n = len(quad)
    def on_segment(a, b, p):
        x1, y1 = a
        x2, y2 = b
        xp, yp = p
        cross = (x2 - x1) * (yp - y1) - (y2 - y1) * (xp - x1)
        if cross != 0:
            return False
        if min(x1, x2) <= xp <= max(x1, x2) and min(y1, y2) <= yp <= max(y1, y2):
            return True
    inside = False
    for i in range(n):
        a = quad[i]
        b = quad[(i +1) % n]
        if on_segment(a, b, p):
            return True
        x1, y1 = a
        x2, y2 = b
        intersect = ((y1 > y) != (y2 > y)) and (x < (x2 -x1) * (y - y1) / (y2 - y1) + x1)
        if intersect:
            inside = not inside
    return inside
x1, y1 = map(float, input().split())
x2, y2 = map(float, input().split())
x3, y3 = map(float, input().split())
x4, y4 = map(float, input().split())
xp , yp = map(float, input().split())
quad = [(x1, y1), (x2, y2), (x3, y3), (x4, y4)]
p = (xp, yp)
print(point_in_quadrilateral(quad, p))
```

- Nếu a == 0 và b == 0 và c != 0, phương trình vô nghiệm. 


```python
def quadratic_roots_count(a, b,c):
    eps = 1e-9
    if abs(a) > eps:
        d = b**2 -4 * a * c
        if d > eps:
            return 2
        elif abs(d) <= eps:
            return 1
        else:
            return 0
    elif abs(b) > eps:
        return 1
    else:
        if abs(c) <= eps:
            return float('inf')
        else:
            return 0
a, b, c = map(float, input().split())
print(quadratic_roots_count(a, b, c))
    
    
```

```python
def is_perfect_square(a, b, c):
    eps = 1e-9
    if a <= 0 or b <= 0 or c <= 0:
        return False
    sides = sorted([a, b, c])
    x, y, z = sides
    if x + y <= z:
        return False
    return abs(x**2 + y**2 - z**2) < eps
a, b ,c = map(int, input().split())
print(is_perfect_square(a, b, c))
```

33: Yêu cầu: Viết hàm compute(a, b, op) tính a (+|-||/) b tùy theo op.
- Đầu vào: các số a, b và ký tự op thuộc {'+', '-', '', '/'}.
- Đầu ra: kết quả số học.
- Yêu cầu thêm:
- Nếu op là '/' và b == 0, phát sinh ValueError.
- Nếu op không hợp lệ, phát sinh ValueError. 


```python
def compute(a, b, op):
    if op == '+':
        return a + b
    elif op == '-':
        return a - b
    elif op == '*':
        return a * b
    elif op == '/':
        if b == 0:
            return ValueError('Khong chia duoc cho 0')
        return a / b
    else:
        raise ValueError('Toan tu khong hop le')
a, b = map(float, input().split())
op = input()
print(compute(a, b, op))
```

35: Yêu cầu: Viết hàm divide_with_remainder(dividend, divisor) trả về bộ (q, r) trong đó q là thương nguyên và r là số dư sao
cho dividend = divisorq + r với 0 <= |r| < |divisor| (dùng ngữ nghĩa của Python với // và %).
- Đầu vào: số nguyên dividend, divisor (divisor khác 0).
- Đầu ra: bộ hai số nguyên (q, r).
- Ví dụ: divide_with_remainder(5, 3) -> (1, 2)


```python
def devide_with_remainder(dividend, divisor):
    if divisor == 0:
        raise ValueError('so chia khong duoc bang 0')
    q = dividend // divisor
    r = dividend % divisor
    return (q, r)
dividend, divisor = map(int, input().split())
print(devide_with_remainder(dividend, divisor))
```

41: Viết hàm trả về ký tự đứng sau ký tự x n vị trí trong bảng chữ cái.
Giữ nguyên chữ hoa/chữ thường và quay vòng trong phạm vi A..Z hoặc a..z

```python
def next_char(x, n):
    if len(x) != 1 or not x.isalpha():
        raise ValueError('Phai la dung 1 ki tu chu cai')
    if x.isupper():
        base = ord('A')
        return chr((ord(x) - base + n) % 26 + base)
    else:
        base = ord('a')
        return chr((ord(x) - base + n) % 26 + base)
x = input()
n = int(input())
print(next_char(x, n))
```

43: Viết hàm tính diện tích tam giác có ba cạnh a, b, c (dùng công thức Heron). Phát sinh ValueError nếu ba cạnh không thể
tạo thành tam giác. 


```python
import math
def triangle_area(a, b, c):
    if a <= 0 or b <= 0 or c <= 0:
        return False
    if not (a + b > c) and (a + c > b) and (b + c > a):
        raise ValueError('Khong tao thanh tam giac')
    s = (a + b + c) / 2
    area = math.sqrt(s * (s - a) * (s - b) * (s -c))
    return area
a, b, c = map(float, input().split())
print(triangle_area(a, b, c))
```

44: Viết hàm trả về tích các số nguyên từ 1 đến n (giai thừa). Với n=0, trả về 1.

```python
def factorial(n):
    if n < 0:
        raise ValueError('n phai la so nguyen khong am')
    result = 1
    for i in range(1, n+1):
        result *= i
    return result
n = int(input())
print(factorial(n))
```

45: Viết hàm trả về số lượng số nguyên dương k sao cho 1 <= k < n và k > (n - 1) / 2. (Ví dụ: n=9 -> 4). 

```python
def count_k(n):
    if n <= 1:
        return 0
    cnt = 0
    for k in range(1, n):
        if k > (n - 1) / 2:
            cnt += 1
    return cnt
n = int(input())
print(count_k(n))
```
