W8A1: [Cylinder]
Xây dựng lớp Cylinder (hình trụ) gồm: bán kính và chiều cao
(kiểu số thực) với hai phương thức:tính diện tích bề mặt (diện
tích toàn phần) và thể tích của hình trụ
Đầu vào
Đâu vào từ bàn phím gồm một dòng duy nhất chứa bán kính và
chiều cao của hình trụ, cách nhau bởi một dấu cách.
Đầu ra
In ra màn hình diện tích bề mặt và thể tích của hình trụ (sử
dụng lại những hàm vừa viết, π=3.14).
(Kết quả làm tròn đến 2 chữ số sau dấu phẩy). 


```python
class Cylinder:
    def __init__(self, bankinh, chieucao):
        self.bk = bankinh
        self.cc = chieucao

    def getArea(self):
        return 2*3.14*self.bk**2 + 2*3.14*self.bk*self.cc
    
    def getVolume(self):
        return (3.14*self.bk**2) * self.cc
    
    def __str__(self):
        return f"({self.bk}, {self.cc})"
    
r, h = map(float, input().split())
Cylinder1 = Cylinder(r, h)
print(Cylinder1)
print(f"Dien tich be mat: {Cylinder1.getArea():.2f}")
print(f"The tich: {Cylinder1.getVolume():.2f}")
```

W8A2: [Date]
Xây dựng lớp Date biểu diễn ngày tháng năm với hai phương
thức kiểm tra xem một ngày tháng nào đó có hợp lệ hay không
và phương thức tính ngày kế tiếp của ngày tháng năm nào đó
nếu nó là hợp lệ
Nhập vào một ngày tháng năm, (ngày tháng năm biểu diễn
dưới dạng dd/mm/yyyy). Nếu ngày tháng năm không tồn tại, in
ra dòng chữ INVALID. Nếu là hợp lệ in ra ngày kế tiếp của nó.
Đầu vào
Đầu vào từ bàn phim chứa dữ liệu ngày, tháng, năm dưới dạng
dd/mm/yyyy
.
Đầu ra
In ra màn hình ngày kế tiếp của ngày vừa nhập dưới dạng
dd/mm/yyyy
. Nếu ngày tháng không tồn tại, in ra INVALID 


```python
class My_Date:
    def __init__(self, d, m, y):
        self.dd = d
        self.mm = m
        self.yyyy = y 

    def checkValid(self):
        max_day = -1
        match self.mm:
            case 1|3|5|7|8|10|12:
                max_day = 31
            case 4|6|9|11:
                max_day = 30
            case 2:
                if (self.yyyy % 4 == 0 and self.yyyy % 100 != 0) or (self.yyyy % 400 != 0):
                    max_day = 29
                else:
                    max_day = 28
        return 1 <= self.dd <= max_day and self.yyyy > 0
    
    def getNextDay(self):
        next_day = My_Date(self.dd + 1, self.mm, self.yyyy)
        if next_day.checkValid():
            return next_day
        next_day.dd = 1
        next_day.mm += 1
        if next_day.checkValid():
            return next_day
        next_day.mm = 1
        next_day.yyyy += 1
        return next_day
    
    def __str__(self):
        return f"{self.dd:02d}/{self.mm:02d}/{self.yyyy}"
d, m, y = map(int, input().split("/"))
My_Date1 = My_Date(d, m, y)
print(My_Date1)

if My_Date1.checkValid():
    next_day = My_Date1.getNextDay()
    print(next_day)
else: 
    print('INVALID')
```

W8A3: [Point]
Xây dựng lớp điểm với hai thuộc tính x, y cùng các phương
thức kiểm tra xem có phải gốc tọa dộ không, có nằm trên trục
tung không, có nằm trên trục hoành không, tính khoảng cách từ
điểm đó tới một điểm khác
Đầu vào:
Nhập vào tọa độ một điểm M bất kỳ
Đầu ra:
Thông báo M có phải gốc tọa độ không hoặc có nằm trên trục
tung hoặc trục hoành không, khoảng cách từ M tới gốc tọa độ
là bao nhiêu (lấy 2 chữ số phần thập phân) 


```python
import math
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def is_origin(self):
        return self.x == 0 and self.y == 0
    def is_on_y_axis(self):
        return self.x == 0
    def is_on_x_axis(self):
        return self.y == 0
    def distance_to(self, other):
        return math.sqrt((self.x - other.x) ** 2 + (self.y - other.y) ** 2)
x = float(input("Nhập x: "))
y = float(input("Nhập y: "))
M = Point(x, y)
O = Point(0, 0)
if M.is_origin():
    print("M là gốc tọa độ.")
else:
    print("M không phải gốc tọa độ.")
if M.is_on_y_axis():
    print("M nằm trên trục tung.")
else:
    print("M không nằm trên trục tung.")
if M.is_on_x_axis():
    print("M nằm trên trục hoành.")
else:
    print("M không nằm trên trục hoành.")
print(f"Khoảng cách từ M đến gốc tọa độ: {M.distance_to(O):.2f}")
```

```python
class Calculator:
    def __init__(self, x, y):
        self.x = x
        self.y = y 
    def __add__(self):
        return self.x + self.y
    def __sub__(self):
        return self.x - self,y
    def __mul__(self):
        return self.x * self.y
    def __truediv__(self):
        if self.y != 0:
            return self.x / self.y
        return None
    def __pow__(self):
        return self.x ** self.y
    def __mod__(self):
        if self.y != 0:
            return self.x % self.y
        return None
    def set_number(self, x, y):
        self.x = x
        self.y = y
x, y = map(float, input().split())
calc = Calculator(x, y)
while True:
    choice = int(input())
    if choice == 1:
        print(f"{x + y:.2f}")
    elif choice == 2:
        print(f"{x - y:.2f}")
    elif choice == 3:
        print(f"{x * y:.2f}")
    elif choice == 4:
        print(f"{x / y:.2f}")
    elif choice == 5:
        print(f"{x ** y:.2f}")
    elif choice == 6:
        print(f"{x % y:.2f}")
    elif choice == 7:
        x, y = map(float, input().split())
        print(calc.set_number(x, y))
    else:
        print('Chon sai')
    exit_choice = input("Ban co muon thoat")
    if exit_choice.lower() in ["yes", "YES", "Yes"]:
        print("Stop")
        break
        
        
```

W8A5: [ShoppingCart]
Xây dựng lớp ShoppingCart (giỏ hàng) để quản lý các sản
phẩm của khách. Dùng danh sách để lưu các sản phẩm, mỗi
sản phẩm là một từ điển có name (tên) và price(giá).
Với các phương thức sau:
1. Khởi tạo với danh sách rỗng
2. Thêm vào giỏ hàng một sản phẩm mới
3. Xóa khỏi giỏ một sản phẩm nào đó (tìm theo tên để xóa)
4. Kiểm tra xem giỏ hàng có rỗng không
5. Tính tổng tiền giỏ hàng
6. Hiển thị toàn bộ sản phẩm trong giỏ
7. Xóa toàn bộ giỏ hàng
Tạo menu cho người dùng nhập lựa chọn truy vấn từ 2 tới 7.
Chương trình dừng khi người dùng nhập "yes" cho câu hỏi "Có
muốn thoát khỏi chương trình không?" 

```python
class ShoppingCart:
    def __init__(self):
        self.products = []
    def add_product(self, name, price):
        self.products.append({"name": name, "price": price})
    def remove_product(self, name):
        for product in self.products:
            if product["name"] == name:
                self.products.remove(product)
                print("Đã xóa sản phẩm:", name)
                return
        print("Không tìm thấy sản phẩm!")
    def is_empty(self):
        return len(self.products) == 0
    def total_price(self):
        total = 0
        for product in self.products:
            total += product["price"]
        return total
    def display_products(self):
        if self.is_empty():
            print("Giỏ hàng rỗng!")
        else:
            print("\nDanh sách sản phẩm:")
            for product in self.products:
                print(f"Tên: {product['name']}, Giá: {product['price']}")
    def clear_cart(self):
        self.products.clear()
        print("Đã xóa toàn bộ giỏ hàng!")
cart = ShoppingCart()
while True:
    choice = int(input())
    if choice == 2:
        name = input("Nhập tên sản phẩm: ")
        price = float(input("Nhập giá sản phẩm: "))
        cart.add_product(name, price)
    elif choice == 3:
        name = input("Nhập tên sản phẩm cần xóa: ")
        cart.remove_product(name)
    elif choice == 4:
        if cart.is_empty():
            print("Giỏ hàng đang rỗng.")
        else:
            print("Giỏ hàng không rỗng.")
    elif choice == 5:
        print("Tổng tiền:", cart.total_price())
    elif choice == 6:
        cart.display_products()
    elif choice == 7:
        cart.clear_cart()
    else:
        print("Lựa chọn không hợp lệ!")
    thoat = input("Có muốn thoát khỏi chương trình không? (yes/no): ")
    if thoat.lower() == "yes":
        break
```

Xây dựng lớp Rectangle(w, h) với các phương thức:
area() trả về diện tích.
perimeter() trả về chu vi.
scale(k) phóng to hình chữ nhật theo hệ số k (nhân cả w và h),
không trả về gì.
Input
w h k (số thực dương)
Output
area perimeter sau khi phóng to (làm tròn 2 chữ số thập phân). 


```python
class Rectangle:
    def __init__(self, w, h):
        self.w = w
        self.h = h
    def area(self):
        return self.w * self.h
    def perimeter(self):
        return 2 * (self.w + self.h)
    def scale(self, k):
        self.w *= k
        self.h *= k
w, h, k = map(float, input().split())
r = Rectangle(w, h)
r.scale(k)
print(f"{r.area():.2f} {r.perimeter():.2f}")
    
```

W8A7: Yêu cầu
Lớp Fraction(num, den) (tử, mẫu nguyên, mẫu ≠ 0) với:
Rút gọn phân số trong __init__ (quy ước mẫu > 0).
Các phương thức add/sub/mul/div trả về phân số đã rút gọn.
__str__() in dạng a/b.
Input
a b op c d (tương ứng a/b (op) c/d, op thuộc {+,-,*,/})
Output
Phân số kết quả đã rút gọn 


```python
from math import gcd
class Fraction:
    def __init__(self, num, den):
        g = gcd(num, den)
        num //= g
        den //= g
        if den < 0:
            num = -num
            den = -den
        self.num = num
        self.den = den
    def add(self, other):
        return Fraction(self.num * other.den + other.num * self.den, 
                        self.den * other.den)
    def sub(self, other):
        return Fraction(self.num * other.den - other.num * self.den,
                        self.den * other.den)
    def mul(self, other):
        return Fraction(self.num * other.num, self.den * other.den)
    def div(self, other):
        return Fraction(self.num * other.den, self.den * other.num)
    def __str__(self):
        return f"{self.num} / {self.den}"
a, b, op, c, d = input().split()
a, b, c, d = int(a), int(b), int(c), int(d)
f1 = Fraction(a, b)
f2 = Fraction(c, d)
if op == '+':
    result = f1.add(f2)
elif op == '-':
    result = f1.sub(f2)
elif op == '*':
    result = f1.mul(f2)
elif op == '/':
    result = f1.div(f2)
print(result)
        
```

W8A8: Yêu cầu
Lớp BankAccount(owner, balance) với:
deposit(x), withdraw(x); nếu rút quá số dư thì bỏ qua giao dịch.
__str__() in owner: balance.
Input
Dòng 1: owner initial_balance
Dòng 2: n (số lệnh)
Tiếp theo n dòng, mỗi dòng: DEPOSIT x hoặc WITHDRAW x.
Output
Số dư cuối cùng (in số thực, 2 chữ số). 

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance
    def deposit(self, x):
        self.balance += x
    def withdraw(self, x):
        if x <= self.balance:
            self.balance -= x
    def __str__(self):
        return f"{self.owner}: {self.balance:.2f}"
owner, initial_balance = input().split()
account = BankAccount(owner, float(initial_balance))
n = int(input())
for _ in range(n):
    command, amount = input().split()
    amount = float(amount)
    if command == 'DEPOSIT':
        account.deposit(amount)
    elif command == 'WITHDRAW':
        account.withdraw(amount)
print(f"{account.balance:.2f}")
```

W8A9: Yêu cầu
Lớp Student(name) với:
add_score(subject, score)
avg() trả về điểm trung bình.
rank() trả về "Excellent" nếu avg ≥ 8, "Good" nếu 6.5–<8,
"Average" nếu 5–<6.5, ngược lại "Poor".
Input
Dòng 1: name
Dòng 2: n
n dòng tiếp: subject score (score là số thực 0–10)
Output
name avg rank (avg 2 chữ số). 

```python
class Student:
    def __init__(self, name):
        self.name = name
        self.scores = {}
    def add_score(self, subject, score):
        self.scores[subject] = score
    def avg(self):
        return sum(self.scores.values()) / len(self.scores)
    def rank(self):
        a = self.avg()
        if a >= 8:
            return 'Excellent'
        elif 6.5 <= a < 8:
            return 'Good'
        elif 5 <= a < 6.5:
            return 'Poor'
name = input()
student = Student(name)
n = int(input())
for _ in range(n):
    subject, score = input().split()
    student.add_score(subject, float(score))
print(student.name, f"{student.avg():.2f}", student.rank())
```

W8A10:  Yêu cầu
Lớp Book(title, author, year).
Lớp Library() với:
add(book)
count_by_author(name)
find_by_year(y) trả về số sách xuất bản năm y.
Input
Dòng 1: n (số thao tác)
n dòng:
ADD title;author;year
COUNT author (đếm theo tác giả)
COUNTYEAR year (đếm theo năm)
Output
Mỗi lệnh COUNT/COUNTYEAR in một dòng kết quả

```python
class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year
class Library:
    def __init__(self):
        self.books = []
    def add(self, book):
        self.books.append(book)
    def count_by_author(self, name):
        count = 0
        for book in self.books:
            if book.author == name:
                count += 1
        return count
    def find_by_year(self, y):
        count = 0
        for book in self.books:
            if book.year == y:
                count += 1
        return count
library = Library()
n = int(input())
for _ in range(n):
    line = input()
    if line.startswith('ADD'):
        data = line[4:]
        title, author, year = data.split(";")
        library.add(Book(title, author, int(year)))
    elif line.startswith('COUNTYEAR'):
        year = int(line.split()[1])
        print(library.find_by_year(year))
    elif line.startswith('COUNT'):
        author = line[6:]
        print(library.count_by_author(author))

```

W8A11: Bạn được cho bốn điểm A, B, C và D trong hệ tọa độ Đề-các 3
chiều.
Nhiệm vụ của bạn là tính và in ra góc giữa hai mặt phẳng tạo
bởi các điểm:
mặt phẳng (A, B, C)
mặt phẳng (B, C, D)
Góc cần in ra được ký hiệu là PHI, và phải dùng độ (degrees),
không dùng radian.
Dạng Input:
Một dòng input chứa 3 số thực, ngăn cách bằng dấu cách, biểu
diễn:
tọa độ X tọa độ Y tọa độ Z của một điểm.
Dạng Output:
In ra góc giữa hai mặt phẳng, chính xác tới hai chữ số thập
phân. 

```python
from math import sqrt, acos, degrees
class Point:
    def __init__(self, x, y, z):
        self.x = x
        self.y = y
        self.z = z
def vector (a, b):
        return (b.x - a.x, 
                b.y - a.y, 
                b.z - a.z)
def cross(u, v):
        return (u[1] * v[2] - u[2] * v[1],
                u[2] * v[0] - u[0] * v[2],
                u[0] * v[1] - u[1] * v[0])
def dot(u, v):
        return u[0] * v[0] + u[1] * v[1] + u[2] * v[2]
def norm(v):
        return sqrt(dot(v, v))
A = Point(*map(float, input().split()))
B = Point(*map(float, input().split()))
C = Point(*map(float, input().split()))
D = Point(*map(float, input().split()))
AB = vector(A, B)
AC = vector(A, C)
BC = vector(B, C)
BD = vector(B, D)
n1 = cross(AB, AC)
n2 = cross(BC, BD)
cos_phi = abs(dot(n1, n2)) / (norm(n1) * norm(n2))
phi = degrees(acos(cos_phi))
print(f"{phi:.2f}")
    
```

W8A12: [Complex]
Khai báo kiểu dữ liệu biểu diễn số phức \(ComplexNumber\)
bao gồm \(2\) biến \(realPart\) và \(imaginaryPart\) kiểu
\(double\) biểu diễn phần thực và phần ảo của một số phức.
Viết hàm khởi tạo ComplexNumber(double initRealPart, double
initImaginaryPart) để khởi tạo các giá trị phần thực và phần ảo
của một số phức và hàm ComplexNumber
multiply(complexNumber other) trả về tích của số phức hiện tại
và một số phức khác.
Đầu vào
Đầu vào từ bàn phím gồm hai dòng, mỗi dòng gồm hai số thực
cách nhau bởi một dấu cách biểu diễn phần thực và phần ảo
của một số phức. Các số trong đầu vào có giá trị tuyệt đối
không vượt quá \(100\).
Đầu ra
In ra màn hình hai số biểu diễn phần thực và phần ảo của tích
hai số phức.


```python
class ComplexNumber:
    def __init__(self, initrealPart, initImaginaryPart):
        self.realPart = initrealPart
        self.imaginaryPart = initImaginaryPart
    def multiply(self, other):
        real = self.realPart * other.realPart - self.imaginaryPart * other.imaginaryPart
        imag = self.realPart * other.imaginaryPart + self.imaginaryPart * other.realPart
        return ComplexNumber(real, imag)
a, b = map(int, input().split())
c, d = map(int, input().split())
z1 = ComplexNumber(a, b)
z2 = ComplexNumber(c, d)
result = z1.multiply(z2)
print(result.realPart, result.imaginaryPart)
```

```python
import math
class Triangle:
    def __init__(self, initEdge1, initEdge2, initEdge3):
        self.edge1 = initEdge1
        self.edge2 = initEdge2
        self.edge3 = initEdge3
    def getArea(self):
        a, b, c = self.edge1, self.edge2, self.edge3
        if a + b <= c or a + c <= b or b + c <= a:
            return -1
        p = (a + b + c) / 2
        return sqrt(p * (p - a) * (p - b) * (p - c))
a, b, c = map(int, input().split())
triangle = Triangle(a, b, c)
area = triangle.getArea()
if area == -1:
    print('Invalid')
else:
    print(f"{area:.2f}")
```

Xây dựng lớp Animal với phương thức speak().
Xây dựng hai lớp kế thừa Dog và Cat, mỗi lớp ghi đè speak()
để trả về âm thanh tương ứng:
- Dog -> Woof
- Cat -> Meow
Input
Một dòng chứa tên lớp: Dog hoặc Cat.
Dog Woof
Output
In ra kết quả của speak()

```python
class Animal:
    def __init__(self):
        pass
class Dog(Animal):
    def speak(self):
        return 'Gau'
class Cat(Animal):
    def speak(self):
        return 'Meo'
animal_type = input().strip()
if animal_type == 'Dog':
    animal = Dog()
elif animal_type == 'Cat':
    animal = Cat()
print(animal.speak())

```

W8A15: Yêu cầu
Xây dựng lớp Employee(name, salary).
Xây dựng lớp Manager(name, salary, bonus) kế thừa từ
Employee.
Cả hai lớp đều có phương thức income():
- Employee: trả về salary
- Manager: trả về salary + bonus
Input
Dòng 1: loại đối tượng (Employee hoặc Manager)
Dòng 2: name salary [bonus]
Output
In ra tổng thu nhập của đối tượng.

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = float(salary)
    def income(self):
        return self.salary
class Manager(Employee):
    def __init__(self, name, salary, bonus):
        super().__init__(name, salary)
        self.bonus = float(bonus)
    def income(self):
        return self.salary + self.bonus
obj_type = input().strip()
data = input().split()
if obj_type == 'Employee':
    name, salary = data
    employee = Employee(name, salary)
    print(employee.income())
elif obj_type == 'Manager':
    name, salary, bonus = data
    manager = Manager(name, salary, bonus)
    print(manager.income())
    
```
