W6A16: Nhâ ̣p vào mảng n hàng, m cột số nguyên. In lại mảng ra màn hình như ma ̣
trận hai chiều, định dạng mỗi phần tử 4 vị trí căn phải


```python
def print_matrix(matrix):
    for row in matrix:
        for val in row:
            print(f"{val:4}", end="")
        print()
n, m = map(int, input().split())
matrix = []
for _ in range(n):
    row = list(map(int, input().split()))
    matrix.append(row)
print(print_matrix(matrix))
```

W6A17: Nhập vào mảng vuông 2 chiều n hàng, n cột, in ra đường chéo chính và
đường chéo phụ của mảng

```python
def diagonals(matrix):
    n = len(matrix)
    main_diag = [matrix[i][i] for i in range(n)]
    second_diag = [matrix[i][n - 1 - i] for i in range(n)]
    print('Duong cheo chinh:', main_diag)
    print('Duong cheo phu:', second_diag)
n = int(input())
matrix = []
for _ in range(n):
    row = list(map(int, input().split()))
    matrix.append(row)
print(diagonals(matrix)) 
```

W6A18: Nhâ ̣p vào mảng n hàng, m cột số nguyên và số nguyên k (n, m, k nhập trên ̣
cùng một dòng, cách nhau dấu cách). Tính và in ra tổng các phần tử trên cột
k (hàng k) 


```python
def tong_cot_k():
    n, m, k = map(int, input().split())
    a = []
    for _ in range(n):
        row = list(map(int, input().split()))
        a.append(row)
    k -= 1
    s = 0
    for i in range(n):
        s += a[i][k]
    print(s)
print(tong_cot_k())
```

W6A19: Viết hàm đưa vào 1 list số nguyên, tìm và trả về vị trí có giá trị lớn nhất
trong list


```python
def max_pos(a):
    max_val = a[0]
    pos = 0
    for i in range(1, len(a)):
        if a[i] > max_val:
            max_val = a[i]
            pos = i
    return pos
a = list(map(int, input().split()))
print(max_pos(a))
```

W6A20: Viết hàm đưa vào một list số nguyên và một số nguyên dương k. Hãy tìm và
trả về vị trí của phần tử đầu tiên có giá trị k trong list, nếu không có thì trả về
-1

```python
def first_pos(a, k):
    for i in range(len(a)):
        if a[i] == k:
            return i + 1
    return -1
a = list(map(int, input().split()))
k = int(input())
print(first_pos(a, k))
```

W6A21: Nhập một dòng gồm các chuỗi (tách bởi dấu cách). Tạo dictionary ánh xạ
chuỗi -> tuple các vị trí (0-based) mà chuỗi đó xuất hiện

```python
def dictionary_pos():
    words = input().split()
    d = {}
    for i in range(len(words)):
        w = words[i]
        if w not in d:
            d[w] = []
        d[w].append(i)
    for k in d:
        d[k] = tuple(d[k])
    print(d)
dictionary_pos()
```

W6A22: Viết hàm đưa vào 1 list số nguyên, tìm và trả về vị trí có giá trị lớn nhất
trong list

```python
def max_pos(a):
    pos = 0
    for i in range(len(a)):
        if a[i] > a[pos]:
            pos = i
    return pos
a = list(map(int, input().split()))
print(max_pos(a))
```

W6A23: VIết hàm xóa đi các số <x có trong mảng

```python
def nho_hon_x(a, x):
    i = 0
    while i < len(a):
        if a[i] < x:
            a.pop(i)
        else:
            i += 1
    return a
a = list(map(int, input().split()))
x = int(input())
print(nho_hon_x(a, x))
```

W6A24: Viêết hàm đẩy tất cả các số 0 về đầu (về cuối), các số khác vẫn đúng vị trí
trước sau: 0 2 3 2 4 0 0 0 9 => 0 0 0 2 3 2 4 9

```python
def day_0_ve_dau(a):
    zeros = []
    others = []
    for x in a:
        if x == 0:
            zeros.append(x)
        else:
            others.append(x)
    return zeros + others
a = list(map(int, input().split()))
print(day_0_ve_dau(a))
```

W6A25: Cho dãy chỉ gồm N giá trị từ 0 - 9 . Kiểm tra xem giá trị nào không có trong
mảng


```python
def missing_value(a):
    all_digits = set(range(10))
    present = set(a)
    missing = all_digits - present
    return sorted(missing)
a = list(map(int, input().split()))
print(missing_value(a))
```

W6A26: Tính tổng các phần tử cực đại địa phương trong một danh sách, không tính 2
phần tử đầu và cuối


```python
def tong_cuc_dai(a):
    s = 0
    for i in range(1, len(a) - 1):
        if a[i] > a[i - 1] and a[i] > a[i + 1]:
            s += a[i]
    return s
a = list(map(int, input().split()))
print(tong_cuc_dai(a))
```

W6A27: Cho một dãy số nguyên, Tìm và in ra vị trí các số < x (x là giá trị cho trước).
nếu không có in ra -1

```python
def vi_tri_nho_hon_x(a, x):
    res = []
    for i in range(len(a)):
        if a[i] < x:
            res.append(i)
    if len(res) == 0:
        print(-1)
    else:
        print(*res)
a = list(map(int, input().split()))
x = int(input())
print(vi_tri_nho_hon_x(a, x))
```

W6A28: Cho một dãy các số nguyên, kiểm tra xem nó là dãy tăng thực sự hay giảm
thực sự, hay vô danh (không tăng, không giảm)

```python
def check(a):
    tang = True
    giam = True
    for i in range(len(a) - 1):
        if a[i] >= a[i + 1]:
            tang = False
        if a[i] < a[i + 1]:
            giam = False
    if tang:
        print('tang')
    elif giam:
        print('giam')
    else:
        print('vo danh')
a = list(map(int, input().split()))
print(check(a))
```
