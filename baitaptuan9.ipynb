W9A1: Viết hàm merge hai danh sách đã sắp xếp mà không dùng bất cứ thuật toán sắp xếp nào 


```python
def merge_lists(a, b):
    i = j = 0
    result = []
    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            result.append(a[i])
            i += 1
        else:
            result.append(b[j])
            j += 1
    while i < len(a):
        result.append(a[i])
        i += 1
    while j < len(b):
        result.append(b[j])
        j += 1
    return result
a = list(map(int, input().split()))
b = list(map(int, input().split()))
print(merge_lists(a, b))

```

W9A2: Cho ma trận hai chiều nxm có các phần tử được sắp xếp trên dòng tăng dần và trên cột
tăng dần, hãy tìm số lớn thứ k (k < nxm) với thuật toán tốt nhất có thể

```python
import heapq
def kth_largest(matrix, k):
    n = len(matrix)
    m = len(matrix[0])
    heap = []
    visited = set()
    heapq.heappush(heap, (-matrix[n - 1][m - 1], n - 1, m - 1))
    visited.add((n - 1, m - 1))
    for _ in range(k):
        val, r, c = heapq.heappop(heap)
        if r - 1 >= 0 and (r - 1, c) not in visited:
            heapq.heappush(heap, (-matrix[r - 1][c], r - 1, c))
            visited.add((r - 1, c))
        if c - 1 >= 0 and (r, c - 1) not in visited:
            heapq.heappush(heap, (-matrix[r][c - 1], r, c - 1))
            visited.add((r, c - 1))
    return -val
n, m = map(int, input().split())
matrix = [list(map(int, input().split())) for _ in range(n)]
k = int(input())
print(kth_largest(matrix, k))
```

W9A3: Cho một số nguyên dương, hãy thực hiện hoán đổi đúng 2 chữ số để được một số
nguyên nhỏ nhất có thể

```python
def smallest_after_one_swap(n):
    s = list(str(n))
    m = len(s)
    for i in range(m):
        min_digit = s[i]
        pos = -1
        for j in range(i + 1, m):
            if s[j] < min_digit:
                min_digit = s[j]
                pos = j
            elif s[j] == min_digit:
                pos = j
        if pos != -1:
            if i == 0 and s[pos] == '0':
                continue
            s[i], s[pos] = s[pos], s[i]
            return int(''.join(s))
    return n
n = int(input())
print(smallest_after_one_swap(n))
```

W9A4: Cho xâu a gồm các chữ cái thường và in hoa và xâu b chỉ gồm các chữ cái in hoa. Ta có
thể biến đổi xâu a theo 2 bước như sau:
- Chuyển một số chữ cái thường trong xâu a thành chữ cái in hoa (có thể không biến đổi
chữ cái nào)
- Xóa hết tất cả các chữ cái thường trong xâu a.
Cho 2 xâu a,b, xác đính xem có cách biến đổi từ xâu a sang xâu b không. Ghi ra "YES"
nếu có và "NO" nếu ngược lại.
Đầu vào:
Dòng đầu tiên chứa 1 số nguyên Q là số lượng câu hỏi. Tương ứng với mỗi câu hỏi, đưa
ra 2 xâu a,b trên 2 dòng khác nhau.(Q≤10,|a|,|b|≤1000
)
Đầu ra:
Với mỗi câu hỏi, ghi ra kết quả tương ứng trên một dòng. 


```python
def can_transform(a, b):
    n = len(a)
    m = len(b)
    dp = [[False] * (m + 1) for _ in range(n + 1)]
    dp[0][0] = True
    for i in range(1, n + 1):
        if a[i - 1].islower():
            dp[i][0] = dp[i - 1][0]
    for i in range(1, n + 1):
        for j in range(1, m + 1):
            ch = a[i - 1]
            if ch.islower():
                dp[i][j] = dp[i - 1][j]
                if ch.upper() == b[j - 1]:
                    dp[i][j] |= dp[i - 1][j - 1]
            else:
                if ch == b[j - 1]:
                    dp[i][j] = dp[i - 1][j - 1]
    return "YES" if dp[n][m] else "NO"
Q = int(input())
for _ in range(Q):
    a = input().strip()
    b = input().strip()
    print(can_transform(a, b))
```

W9A5: Cho hai mảng số nguyên list1 và list2, viết hàm kiểm tra xem list1 có thuộc list2 hay không
(có phải là dãy con liên tiếp của list2 hay không)? 


```python
def is_sub_list(list1, list2):
    n = len(list1)
    m = len(list2)
    if n > m:
        return False
    for i in range(m - n + 1):
        if list2[i:i + n] == list1:
            return True
    return False
lits1 = list(map(int, input().split()))
list2 = list(map(int, input().split()))
print(is_sub_list(lits1, list2))
```

W9A6: [Hoán vị tiếp theo]
Cho một dãy số gồm n chữ số là một hoán vị của dãy 1,2,...,n. Viết chương trình in ra
hoán vị ở vị trí tiếp theo (danh sách các hoán vị được sắp xếp theo thứ tự từ điển) của
hoán vị đã cho.
Đầu vào từ bàn phím gồm 2 dòng:
- Dòng đầu tiên chứa số nguyên n
- Dòng tiếp theo chứa n số nguyên là một hoán vị của dãy 1,2,...,n (n≤105)
Đầu ra: In ra màn hình hoán vị tiếp theo của hoán vị đã cho. 

```python
def next_permutation(a):
    n = len(a)
    i = n - 2
    while i >= 0 and a[i] >= a[i + 1]:
        i -= 1
    if i >= 0:
        j = n - 1
        while a[j] <= a[i]:
            j -= 1
        a[i], a[j] = a[j], a[i]
    left, right = i + 1, n - 1
    while left < right:
        a[left], a[right] = a[right], a[left]
        left += 1
        right -= 1
    return a
n = int(input())
a = list(map(int, input().split()))
print(*next_permutation(a))
```

W9A7: [CanPlaceFlowers]
Giả sử bạn là nhân viên của một công ty cây xanh, được giao cho chăm sóc một luống
đất. Luống đất đó được chia thành nhiều ô đất nhỏ có kích cỡ bằng nhau. Trước khi được
giao cho bạn chăm sóc, trên luống hoa đã được trồng sẵn một số cây xanh tán rộng. Một
ngày đầu xuân, công ty gửi đến cho bạn một số cây hoa và yêu cầu bạn trồng vào những
vị trí còn trống trong luống hoa. Đây là những cây hoa hướng sáng và chúng không thể
sống được dưới bóng râm của cây khác, đặc biệt là cây tán rộng. Vì vậy, bạn không thể
trồng chúng tại ô bên cạnh những ô đã được trồng cây xanh tán rộng. Và nếu trồng những
cây hoa bên cạnh nhau, chúng sẽ tranh giành phân bón và nước và sau cùng, cả hai cây
sẽ cùng chết.
Viết hàm canPlaceFlowers(flowerbed, k) kiểm tra xem có thể trồng hết số hoa mới được
cấp vào trong luống đất của bạn hay không.
Hàm nhận đầu vào là một danh sách (list) flowerbed, đánh dấu những vị trí trong luống đất
đã được trồng cây xanh trước đó (giá trị 1 ứng với vị trí cây đã được trồng và 0 ứng với vị
trí đất trống) và số lượng hoa được phát thêm k (int). Hàm trả về True nếu có thể trồng hết
toàn bộ k cây hoa vào luống đất và False trong trường hợp ngược lại. 


```python
def canPlaceFlowers(flowerbed, k):
    count = 0
    n = len(flowerbed)
    for i in range(n):
        if (flowerbed[i] == 0 and
            (i == 0 or flowerbed[i - 1] == 0) and
            (i == n - 1 or flowerbed[i + 1] == 0)):
            flowerbed[i] = 1
            count += 1
            if count >= k:
                return True
    return count >= k
flowerbed = list(map(int, input().split()))
k = int(input())
print(canPlaceFlowers(flowerbed, k))
```

W9A8: [Phát hiện đường đi lặp lại của Rô-bốt]
Ban đầu, một con rô-bốt ở vị trí (0, 0) trên một lưới 2D.
Cho một chuỗi `moves` chứa các thao tác di chuyển của rô-bốt.
Các bước di chuyển được biểu diễn bởi các ký tự:
- 'R' (Phải)
- 'L' (Trái)
- 'U' (Lên)
- 'D' (Xuống)
Viết hàm `def is_path_crossing(moves)`:
Hàm nhận đầu vào là một chuỗi `moves`.
Hàm trả về `True` nếu trong quá trình di chuyển, rô-bốt đi vào một ô mà nó đã từng đi qua
trước đó.
Hàm trả về `False` nếu rô-bốt không bao giờ đi lặp lại (luôn đi vào ô mới).
Lưu ý: Vị trí ban đầu (0, 0) được tính là vị trí đầu tiên đã đi qua. 


```python
def is_path_crossing(moves):
    x, y = 0, 0
    visited = {(0, 0)}
    for move in moves:
        if move == 'R':
            x += 1
        elif move == 'L':
            x -= 1
        elif move == 'U':
            y += 1
        elif move == 'D':
            y -= 1
        if (x, y) in visited:
            return True
        visited.add((x, y))
    return False
moves = input().strip()
print(is_path_crossing(moves))
```

W9A9: [HARD - Tìm trung vị của 5 số]
Viết hàm `def Median(nums):`
Hàm nhận đầu vào là một danh sách (list) `nums` được đảm bảo có đúng 5 số nguyên.
Yêu cầu:
Tìm và trả về số trung vị (số lớn thứ ba) của 5 số này.
Ràng buộc:
1. Chỉ được sử dụng tối đa 6 phép so sánh (ví dụ: <, >, <=, >=).
2. KHÔNG ĐƯỢC DÙNG bất kỳ hàm sắp xếp nào (như `sorted()`, `list.sort()`).
Chú ý: Bạn chỉ cần viết hàm `Median` theo yêu cầu.
Phần đọc input và gọi hàm (hàm main) đã có sẵn. 

```python
def Median(nums):
    a, b, c, d, e = nums
    if a > b:
        a, b  = b, a
    if c > d:
        c, d = d, c
    if a > c:
        a, c = c, a
    if b > d:
        b, d = d, b
    if b > c:
        b, c = c, b
    if e < b:
        return b
    elif e > c:
        return c
    else:
        return e
nums = list(map(int, input().split()))
print(Median(nums))
```

W9A10: [Tìm bán kính lò sưởi nhỏ nhất]
Mùa đông đang đến! Bạn được cung cấp vị trí của các ngôi nhà và các lò sưởi trên một
đường thẳng.
Nhiệm vụ của bạn là tìm bán kính r nhỏ nhất cho các lò sưởi sao cho mọi ngôi nhà đều
nằm trong phạm vi sưởi ấm của ít nhất một lò sưởi.
Tất cả các lò sưởi đều có cùng một bán kính r.
Một ngôi nhà tại vị trí h được sưởi ấm bởi một lò sưởi tại vị trí t nếu |h - t| <= r.
Viết hàm `def findRadius(houses, heaters):`
Hàm nhận đầu vào là hai danh sách (list) các số nguyên:
- `houses`: chứa vị trí của các ngôi nhà.
- `heaters`: chứa vị trí của các lò sưởi.
Hàm trả về một số nguyên là bán kính r nhỏ nhất cần thiết.
Lưu ý:
- Số lượng nhà và lò sưởi là một số không âm và không vượt quá 25000.
- Vị trí của các ngôi nhà và lò sưởi là một số không âm và không vượt quá 10^9. 

```python
from bisect import bisect_left
def findRadius(houses, heaters):
    heaters.sort()
    radius = 0
    for house in houses:
        pos = bisect_left(heaters, house)
        left_dist = float('inf')
        right_dist = float('inf')
        if pos > 0:
            left_dist = house - heaters[pos - 1]
        if pos < len(heaters):
            right_dist = heaters[pos] - house
        radius = max(radius, min(left_dist, right_dist))
    return radius
houses = list(map(int, input().split()))
heaters = list(map(int, input().split()))
print(findRadius(houses, heaters))
```

W9A11: Viết hàm factorial(n) tính giai thừa của n bằng đệ quy. 


```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)
n = int(input())
print(factorial(n))
```

```python
def sum_to_n(n):
    if n == 1:
        return 1
    return n + sum_to_n(n - 1)
n = int(input())
print(sum_to_n(n))
```

W9A13: Viết hàm power(base, exp) tính base^exp bằng đệ quy chia để trị.
Yêu cầu:
- Không dùng vòng lặp
- Không dùng hàm có sẵn

```python
def power(base, exp):
    if exp == 0:
        return 1
    half = power(base, exp // 2)
    if exp % 2 == 0:
        return half * half
    else:
        return half * half * base
base, exp = map(int, input().split())
print(power(base, exp))
```

W9A14: VIế
́t hàm tìm số xuất hiện nhiều hơn x lần trong một dãy số (x là số nguyên dương nhập
từ bàn phím). Nếu không có số nào thì in ra "INVALID" 

```python
def find_numbers(arr, x):
    count = {}
    for num in arr:
        count[num] = count.get(num, 0) + 1
    result = []
    for num in count:
        if count[num] > x:
            result.append(num)
    if result:
        print(*result)
    else:
        print("INVALID")
arr = list(map(int, input().split()))
x = int(input())
find_numbers(arr, x)
```

W9A15: VIết haàm kiểm tra dãy số nguyên vào từ bàn phím có phải dãy tăng nghiêm ngặt (giảm
nghiêm ngặt) hay không? 

```python
def is_strictly_increasing(arr):
    for i in range(len(arr) - 1):
        if arr[i] >= arr[i + 1]:
            return False
    return True
def is_strictly_decreasing(arr):
    for i in range(len(arr) - 1):
        if arr[i] <= arr[i + 1]:
            return False
    return True
arr = list(map(int, input().split()))
if is_strictly_increasing(arr):
    print("Tang nghiem ngat")
elif is_strictly_decreasing(arr):
    print("Giam nghiem ngat")
else:
    print("Khong phai")
```
