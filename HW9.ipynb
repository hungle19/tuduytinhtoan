W9A16: Cho hai số nguyên dương, n và m (0 < n, m <= 20 ). .Bạn hãy in ra tất cả các chuỗi nhị phân có
n số 0 và m số 1. In theo thứ tự tăng dần trong từ điển 


```python
def generate(s, zeros, ones):
    if zeros == 0 and ones == 0:
        print(s)
        return
    if zeros > 0:
        generate(s + '0', zeros - 1, ones)
    if ones > 0:
        generate(s + '1', zeros, ones - 1)
n, m = map(int, input().split())
generate("", n, m)
```

W9A17: Viết hàm tìm dãy con tăng nghiêm ngặt dài nhất trong dãy đầu vào từ bàn phím 


```python
def longest_increasing_subarray(arr):
    if not arr:
        return []
    max_start = 0
    max_len = 1
    start = 0
    curr_len = 1
    for i in range(1, len(arr)):
        if arr[i] > arr[i - 1]:
            curr_len += 1
        else:
            if curr_len > max_len:
                max_len = curr_len
                max_start = start
            start = i
            curr_len = 1
    if curr_len > max_len:
        max_len = curr_len
        max_start = start
    return arr[max_start:max_start + max_len]
arr = list(map(int, input().split()))
print(*longest_increasing_subarray(arr))
```

W9A18: Cho dãy N phần tử số nguyên, hãy đếm số cặp có vị trí khác nhau và có tổng bằng k (số nguyên
dương cho vào từ bàn phím) 


```python
def count_pairs(arr, k):
    freq = {}
    count = 0
    for x in arr:
        count += freq.get(k - x, 0)
        freq[x] = freq.get(x, 0) + 1
    return count
arr = list(map(int, input().split()))
k = int(input())
print(count_pairs(arr, k))
```

W9A19: Loại bỏ khỏi danh sách các phần tử trùng lặp, giữ lại đúng 1 lần xuất hiện, in lại theo đúng thứ tự
xuất hiện

```python
def remove_duplicates(arr):
    seen = set()
    result = []
    for x in arr:
        if x not in seen:
            seen.add(x)
            result.append(x)
    return result
arr = list(map(int, input().split()))
print(*remove_duplicates(arr))
```

W9A20: Thực hiện chuẩn hóa chuỗi kí tự:
a. bỏ khoảng trắng đầu cuối
b. các khoảng trắng giữa chỉ còn 1 dấu cách
c. viết hoa chữ cái đầu mỗi từ
KHÔNG DÙNG CÁC HÀM SẴN CỐ

```python
def normalize_string(s):
    result = ""
    new_word = True
    started = False
    for ch in s:
        if ch != ' ':
            started = True
            if new_word and 'a' <= ch <= 'z':
                ch = chr(ord(ch) - ord('a') + ord('A'))
            result += ch
            new_word = False
        else:
            if started and not new_word:
                result += ' '
                new_word = True
    if len(result) > 0 and result[-1] == ' ':
        result = result[:-1]
    return result
s = input()
print(normalize_string(s))
```

W9A21: Tìm phần tử lớn thứ k trong dãy n phần tử bằng thuật toán tốt nhất có thể – Tránh dùng sắp xếp
(dùng phân hoạch của quicksort) 

```python
def partition(arr, left, right):
    pivot = arr[right]
    i = left
    for j in range(left, right):
        if arr[j] > pivot:  
            arr[i], arr[j] = arr[j], arr[i]
            i += 1
    arr[i], arr[right] = arr[right], arr[i]
    return i
def quick_select(arr, left, right, k):
    if left <= right:
        p = partition(arr, left, right)
        if p == k - 1:
            return arr[p]
        elif p > k - 1:
            return quick_select(arr, left, p - 1, k)
        else:
            return quick_select(arr, p + 1, right, k)
def kth_largest(arr, k):
    return quick_select(arr, 0, len(arr) - 1, k)
arr = list(map(int, input().split()))
k = int(input())
print(kth_largest(arr, k))
```

W9A22: Hai chuỗi là anagram nếu chúng có cùng ký tự và số lần xuất hiện. viết CT kiểm tra xem 2 chuỗi
đầu vào có phải là anagram của nhau không 

```python
def is_anagram(s1, s2):
    if len(s1) != len(s2):
        return False
    count = {}
    for ch in s1:
        count[ch] = count.get(ch, 0) + 1
    for ch in s2:
        if ch not in count:
            return False
        count[ch] -= 1
    for value in count.values():
        if value != 0:
            return False
    return True
s1 = input()
s2 = input()
print(is_anagram(s1, s2))
```

W9A23: Cho số nguyên rất lớn (string), và một số nguyên dương k. Tìm số có k chữ số lớn nhất có thể
từ số ban đầu bằng cách xóa bớt đi một số chữ số nào đó mà không thay đổi vị trí của bất kì
chữ số nào. (VD: 78934340999, k = 7 => 9440999 là số lớn nhất có 7 chữ số cần tìm) 


```python
def largest_k_digits(num, k):
    remove = len(num) - k
    stack = []
    for digit in num:
        while stack and remove > 0 and stack[-1] < digit:
            stack.pop()
            remove -= 1
        stack.append(digit)
    while remove > 0:
        stack.pop()
        remove -= 1
    return ''.join(stack[:k])
num = input().strip()
k = int(input())
print(largest_k_digits(num, k))
```

W9A24: Nén chuối kí tự bằng cách đếm số lần xuất hiện của các kí tự liên tiếp nhau và ghi lại ngay sau
kí tự đó. VD aaaaaafdhhhhh =>a6f1d1h5 


```python
def compress_string(s):
    if not s:
        return ""
    result = ""
    count = 1
    for i in range(1, len(s)):
        if s[i] == s[i - 1]:
            count += 1
        else:
            result += s[i - 1] + str(count)
            count = 1
    result += s[-1] + str(count)
    return result
s = input()
print(compress_string(s))
```

W9A25: Giả sử mật khẩu mạnh là một chuỗi kí tự dài ít nhất 8 kí tự gồm ít nhất 1 chữ cái thường, 1 chữ
cái hoa, 1 chữ số và 1 trong các kí tự đặc biệt !, @, #, $. Nhập vào một chuỗi kĩ tự, kiểm tra xem
nó có phải là mật khẩu mạnh không. 

```python
def is_strong_password(password):
    if len(password) < 8:
        return False
    has_lower = False
    has_upper = False
    has_digit = False
    has_special = False
    for ch in password:
        if 'a' <= ch <= 'z':
            has_lower = True
        elif 'A' <= ch <= 'Z':
            has_upper = True
        elif '0' <= ch <= '9':
            has_digit = True
        elif ch in '!@#$':
            has_special = True
    return has_lower and has_upper and has_digit and has_special
password = input()
if is_strong_password(password):
    print("True")
else:
    print("False")
```
