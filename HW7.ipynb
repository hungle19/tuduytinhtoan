W7A16: Cho một mảng số nguyên không âm. Gọi mảng đó là "mảng đẹp" nếu
mỗi phần tử x trong mảng xuất hiện đúng x lần.
Kiểm tra xem mảng có phải là mảng đẹp hay không. Nếu không hãy xóa
đi y số (y là nhỏ nhất có thể) để mảng còn lại thành mảng đẹp.


```python
def min_deletions(arr):
    freq = {}
    for x in arr:
        freq[x] = freq.get(x, 0) + 1
    deletions = 0
    for x, cnt in freq.items():
        if cnt == x:
            continue
        elif cnt > x:
            deletions += cnt - x
        else:
            deletions += cnt
    return deletions
arr = list(map(int, input().split()))
print(min_deletions(arr))
```

W7A17: Tìm phần tử xuất hiện nhiều hơn n/2 lần trong mảng (nếu có) với độ
phức tạp thời gian là O(n), độ phức tạp bộ nhớ là O(1), - Boyer–Moore.
Nếu không có phần tử nào thỏa mãn thì không in gì


```python
def majority_element(arr):
    candidate = None
    count = 0
    for x in arr:
        if count == 0:
            candidate = x
            count = 1
        elif x == candidate:
            count += 1
        else:
            count -= 1
    if arr.count(candidate) > len(arr) // 2:
        print(candidate)
arr = list(map(int, input().split()))
majority_element(arr)
```

W7A18: Sắp xếp có điều kiện: Số chẵn đứng trước, số lẻ đứng sau, chẵn tăng
dần, lẻ giảm dần; ; 


```python
def conditional_sort(arr):
    even = sorted([x for x in arr if x % 2 == 0])
    odd = sorted([x for x in arr if x % 2 != 0], reverse=True)
    return even + odd
arr = list(map(int, input().split()))
print(*conditional_sort(arr))
```

W7A19: Merge 2 mảng đã sắp xếp (có thể tăng dần, có thể giảm dần) mà không
dùng bất cứ thuật toán sắp xếp nào


```python
def merge_sorted_arrays(a, b):
    if not a:
        return b
    if not b:
        return a
    ascending = True
    if len(a) > 1:
        ascending = a[0] <= a[-1]
    elif len(b) > 1:
        ascending = b[0] <= b[-1]
    i = j = 0
    result = []
    while i < len(a) and j < len(b):
        if ascending:
            if a[i] <= b[j]:
                result.append(a[i])
                i += 1
            else:
                result.append(b[j])
                j += 1
        else:
            if a[i] >= b[j]:
                result.append(a[i])
                i += 1
            else:
                result.append(b[j])
                j += 1
    result.extend(a[i:])
    result.extend(b[j:])
    return result
a = list(map(int, input().split()))
b = list(map(int, input().split()))
print(*merge_sorted_arrays(a, b))
```

W7A21: Cho N là ngày, T là tháng năm 2020, Tính xem đó là ngày thứ bao nhiêu
của năm đó nếu đó là ngảy hợp lệ


```python
def day_of_year(day, month):
    days_in_month = [31, 29, 31, 30, 31, 30,
                     31, 31, 30, 31, 30, 31]

    if month < 1 or month > 12:
        return None
    if day < 1 or day > days_in_month[month - 1]:
        return None
    return sum(days_in_month[:month - 1]) + day
N = int(input())
T = int(input())
result = day_of_year(N, T)
if result is not None:
    print(result)
```

W7A22: Cho N tên sinh viên gồm cả họ và tên, tìm xem tên nào hay gặp nhất.
Nếu có nhiều tên cùng số lần gặp thì lấy tên có thứ tự từ điển nhỏ nhất

```python
def most_common_name(names):
    freq = {}
    for full_name in names:
        name = full_name.split()[-1]  # lấy tên
        freq[name] = freq.get(name, 0) + 1
    max_count = max(freq.values())
    result = min(name for name, count in freq.items() if count == max_count)
    return result
n = int(input())
names = [input().strip() for _ in range(n)]
print(most_common_name(names))
```

W7A23: Đếm số lượng tên trong xâu kí tự. Tên được định nghĩa là xâu có độ dài >=2 và có chữ cái đầu viết hoa, các chữ cái khác viết thường, chỉ có kí
tự cái.

```python
def count_names(s):
    count = 0
    for word in s.split():
        if (len(word) >= 2 and
            word.isalpha() and
            word[0].isupper() and
            word[1:].islower()):
            count += 1
    return count
s = input()
print(count_names(s))
```

W7A24:  Cho xâu kí tự, xóa các từ giống nhau liền kề, chỉ giữ lại từ đầu tiên

```python
def remove_adjacent_duplicates(s):
    words = s.split()
    if not words:
        return ""
    result = [words[0]]
    for i in range(1, len(words)):
        if words[i] != words[i - 1]:
            result.append(words[i])
    return " ".join(result)
s = input()
print(remove_adjacent_duplicates(s))
```

W7A25: Tìm số có đúng k chữ số lớn nhất có thể trong một số có n chữ số (coi
đầu vào là một số rất lớn - str). Không được hoán đổi vị trí các chữ số


```python
def largest_k_digits(num, k):
    remove = len(num) - k
    stack = []
    for digit in num:
        while remove > 0 and stack and stack[-1] < digit:
            stack.pop()
            remove -= 1
        stack.append(digit)
    return ''.join(stack[:k])
num = input().strip()
k = int(input())
print(largest_k_digits(num, k))
```

W7A27: Cho một dãy ngoặc ()(){}{}[][]..., kiểm tra xem dãy ngoặc đã cho có hợp lệ
hay không

```python
def is_valid_brackets(s):
    stack = []
    pairs = {')': '(', '}': '{', ']': '['}
    for ch in s:
        if ch in '({[':
            stack.append(ch)
        elif ch in ')}]':
            if not stack or stack.pop() != pairs[ch]:
                return False
    return len(stack) == 0
s = input()
if is_valid_brackets(s):
    print("Hop le")
else:
    print("Khong hop le")
```

W7A28: Bà̀i toán Tìm dãy các hợp số dài nhất liên tiếp không chứa số nguyên
nào <= N (N là số nguyên dương cho trước). Nếu có nhiều dãy cùng độ
dài thì lấy dãy nhỏ nhất (gặp đầu tiên từ trái qua)


```python
import math
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(math.sqrt(n)) + 1):
        if n % i == 0:
            return False
    return True
N = int(input())
best_start = 0
best_len = 0
cur_start = 0
cur_len = 0
for x in range(N + 1, N + 1000):
    if not is_prime(x):      
        if cur_len == 0:
            cur_start = x
        cur_len += 1
    else:
        if cur_len > best_len:
            best_len = cur_len
            best_start = cur_start
        cur_len = 0
if cur_len > best_len:
    best_len = cur_len
    best_start = cur_start
for x in range(best_start, best_start + best_len):
    print(x, end=" ")
```

W7A29: Tìm Dãy con tăng dần liên tục dài nhất của một danh sách nhập vào từ
bàn phím. Lấy dãy đầu tiên nếu có nhiều dãy có số phần tử bằng nhau

```python
def longest_increasing_subarray(arr):
    best_start = 0
    best_len = 1
    cur_start = 0
    cur_len = 1
    for i in range(1, len(arr)):
        if arr[i] > arr[i - 1]:
            cur_len += 1
        else:
            if cur_len > best_len:
                best_len = cur_len
                best_start = cur_start
            cur_start = i
            cur_len = 1
    if cur_len > best_len:
        best_len = cur_len
        best_start = cur_start
    return arr[best_start:best_start + best_len]
arr = list(map(int, input().split()))
print(longest_increasing_subarray(arr))
```

W7A30: Tìm Dãy con tăng dần KHÔNG liên tục dài nhất của một danh sách nhập
vào từ bàn phím 

```python
arr = list(map(int, input().split()))
n = len(arr)
dp = [1] * n
parent = [-1] * n
for i in range(n):
    for j in range(i):
        if arr[j] < arr[i] and dp[j] + 1 > dp[i]:
            dp[i] = dp[j] + 1
            parent[i] = j
max_len = max(dp)
idx = dp.index(max_len)
lis = []
while idx != -1:
    lis.append(arr[idx])
    idx = parent[idx]
lis.reverse()
print(lis)
```
