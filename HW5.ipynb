W5A16:Viết hàm kiểm tra một số có phải số Amstronhg hay không? Hàm trả về kiểu bool, có tham số là một số
nguyên x


```python
def is_armstrong(x):
    n = len(str(x))
    total = 0
    temp = x
    while temp > 0:
        digit = temp % 10
        total += digit ** n
        temp //= 10
    return total == x
x = int(input())
print(is_armstrong(x))
```

W5A17: Viết hàm kiêểm tra một số có phải số chính phương hay không? Hàm trả về kiểu bool, có tham số là một số
nguyên x

```python
def is_perfect_square(x):
    if x < 0:
        return False
    sqrt_x = int(x ** 0.5)
    return sqrt_x * sqrt_x == x
x = int(input())
print(is_perfect_square(x))
```

W5A18: VIết hàm kiểm tra một số có phải số đối xứng không, xử lý trên dữ liệu kiểu số nguyên.

```python
def is_palindrome_number(x):
    org = x
    rev= 0
    while x > 0:
        digit = x % 10
        rev = rev * 10 + digit
        x //= 10
    return org == rev
x = int(input())
print(is_palindrome_number(x))
```

W5A19: VIết hàm kiểm tra một số có phải số kỳ diệu hay không, xử lý trên dữ liệu kiểu số nguyên.

```python
def is_magic_number(x):
    while x > 9:
        total = 0
        while x > 0:
            total += x % 10
            x //= 10
        x = total
    return x == 1
x = int(input())
print(is_magic_number(x))
```

W5A20: Viết hàm detect_number_type(s) trả về 'int' nếu s là số nguyên, 'float' nếu là số thực, ngược lại 'neither'.


```python
def detect_number_type(s):
    try:
        int(s)
        return 'int'
    except:
        try:
            float(s)
            return 'float'
        except:
            return 'neither'
s = input()
print(detect_number_type(s))
```

W5A24: Viết hàm are_collinear(x1,y1,x2,y2,x3,y3) kiểm tra ba điểm có thẳng hàng hay không.


```python
def are_collinear(x1, y1, x2, y2, x3, y3):
    return (y2 - y1) * (x3 - x1) == (y3 - y1) * (x2 - x1)
x1, y1, x2, y2, x3, y3 = map(float, input().split())
print(are_collinear(x1, y1, x2, y2, x3, y3))
```

W5A25: Viết hàm point_in_quadrilateral(quad, p) kiểm tra điểm p có nằm trong hoặc trên biên của tứ giác quad hay
không. Trong đó: quad là một tứ giác (4 đỉnh – 4 điểm), p là một điểm cần kiểm tra. VD: quad = [(0,0), (4,0), (4,3),
(0,3)], p = (2,1)


```python
def area(x1, y1, x2, y2, x3, y3):
    return abs((x1*(y2-y3) + x2*(y3-y1) + x3*(y1-y2)) / 2)
def point_in_quadrilateral(quad, p):
    x, y = p
    x1, y1 = quad[0]
    x2, y2 = quad[1]
    x3, y3 = quad[2]
    x4, y4 = quad[3]
    s_quad = area(x1,y1,x2,y2,x3,y3) + area(x1,y1,x3,y3,x4,y4)
    s_point = (
        area(x,y,x1,y1,x2,y2) +
        area(x,y,x2,y2,x3,y3) +
        area(x,y,x3,y3,x4,y4) +
        area(x,y,x4,y4,x1,y1)
    )
    return s_quad == s_point
quad = []
for i in range(4):
    x, y = map(float, input().split())
    quad.append((x, y))
px, py = map(float, input().split())
p = (px, py)
print(point_in_quadrilateral(quad, p))
```

W5A27: Viết hàm trả về số lượng chữ số khác 0 trong n số nguyên nhập vào từ bàn phím (một list)

```python
def count_non_zero_digits(lst):
    count = 0
    for number in lst:
        number = abs(number)
        if number == 0:
            continue
        while number > 0:
            digit = number % 10
            if digit != 0:
                count += 1
            number //= 10
    return count
lst = list(map(int, input().split()))
print(count_non_zero_digits(lst))
```

W5A28:  Viết hàm tính góc ∠ABC (độ) từ tọa độ A(x1,y1), B(x2,y2), C(x3,y3).


```python
import math
def angle_ABC(x1, y1, x2, y2, x3, y3):
    BAx = x1 - x2
    BAy = y1 - y2
    BCx = x3 - x2
    BCy = y3 - y2
    tich_vo_huong = BAx * BCx + BAy * BCy
    length_BA = math.sqrt(BAx**2 + BAy**2)
    length_BC = math.sqrt(BCx**2 + BCy**2)
    cos_angle = tich_vo_huong / (length_BA * length_BC)
    cos_angle = max(-1, min(1, cos_angle))
    angle = math.degrees(math.acos(cos_angle))
    return angle
x1, y1 = map(float, input().split())
x2, y2 = map(float, input().split())
x3, y3 = map(float, input().split())
print(angle_ABC(x1, y1, x2, y2, x3, y3))
```

W5A30: Viết hàm tính hệ số nhị thức C(n, k).

```python
def factorial(n):
    result = 1
    for i in range(1, n+1):
        result *= i
    return result
def C(n, k):
    return factorial(n) // (factorial(k) * factorial(n - k))
n, k = map(int, input().split())
print(C(n, k))
```

W5A34: Viết hàm sum_1_to_n(n) trả về tổng các số x từ 1 đến n, nếu x có các chữ số đôi một khác nhau
Nếu n ≤ 0 trả về 0.

```python
def different_digits(x):
    s = str(x)
    return len(s) == len(set(s))
def sum_1_to_n(n):
    if n <= 0:
        return 0
    total = 0
    for x in range(1, n+1):
        if different_digits(x):
            total += x  
    return total
n = int(input())
print(sum_1_to_n(n))
```
